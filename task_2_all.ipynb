### 9. Task - 1741: Find Total Time Spent by Each Employee - Accept

```python
data = [['1', '2020-11-28', '4', '32'], ['1', '2020-11-28', '55', '200'], ['1', '2020-12-3', '1', '42'], ['2', '2020-11-28', '3', '33'], ['2', '2020-12-9', '47', '74']]
employees = pd.DataFrame(data, columns=['emp_id', 'event_day', 'in_time', 'out_time']).astype({'emp_id':'Int64', 'event_day':'datetime64[ns]', 'in_time':'Int64', 'out_time':'Int64'})
```

```python
df_view_employees = employees
df_view_employees['dif_time'] = df_view_employees['out_time'] - df_view_employees['in_time']
df_view_employees = df_view_employees.groupby(by=['emp_id', 'event_day'])['dif_time'].sum()
df_view_employees = df_view_employees.reset_index()
df_view_employees = df_view_employees.rename(columns={'event_day':'day','dif_time':'total_time'})
df_view_employees = df_view_employees.filter(items=['day','emp_id','total_time'])
print(df_view_employees)
```

### 12. Task - 1907: Count Salary Categories -Accept 02.12.2023

```python
data = [[3, 108939], [2, 12747], [8, 87709], [6, 91796]]
accounts = pd.DataFrame(data, columns=['account_id', 'income']).astype({'account_id':'Int64', 'income':'Int64'})
```

```python
data_view_accounts = [['Low Salary',0],['Average Salary',0],['High Salary',0]]
df_view_accounts = pd.DataFrame(data_view_accounts, columns=['category','accounts_count']).astype({'category':'object', 'accounts_count':'Int64'})

df_view_accounts.loc[df_view_accounts.category =='Low Salary','accounts_count'] = accounts.loc[accounts['income'] < 20000].count().values[1]
df_view_accounts.loc[df_view_accounts.category =='Average Salary','accounts_count'] = accounts.loc[(accounts['income'] >= 20000) & (accounts['income'] <= 50000)].count().values[1]
df_view_accounts.loc[df_view_accounts.category =='High Salary','accounts_count'] = accounts.loc[accounts['income'] > 50000].count().values[1]

print(df_view_accounts)
```

### 13. Task - 184: Department Highest Salary -Accept 04.12.2023

```python
data = [[1, 'Joe', 70000, 1], [2, 'Jim', 90000, 1], [3, 'Henry', 80000, 2], [4, 'Sam', 60000, 2], [5, 'Max', 90000, 1]]
employee = pd.DataFrame(data, columns=['id', 'name', 'salary', 'departmentId']).astype({'id':'Int64', 'name':'object', 'salary':'Int64', 'departmentId':'Int64'})
data = [[1, 'IT'], [2, 'Sales']]
department = pd.DataFrame(data, columns=['id', 'name']).astype({'id':'Int64', 'name':'object'})
```

```python
df_view_emp_gut =employee.merge(department,how="left",left_on='departmentId',right_on='id')
df_view_emp_gut = df_view_emp_gut.rename(columns={'departmentId':'depid','name_y':'Department','name_x':'Employee','salary':'Salary'})
df_view_emp_gut = df_view_emp_gut.filter(items=['depid','Department','Employee','Salary'])

df_view_emp_gut = df_view_emp_gut.loc[df_view_emp_gut.Salary.eq(df_view_emp_gut.groupby('depid').Salary.transform('max'))]
df_view_emp_gut = df_view_emp_gut.filter(items=['Department','Employee','Salary'])
print(df_view_emp_gut)
```

### 14. Task - 585: Investments in 2016 -Accept 05.12.2023

```python
data = [[1, 10, 5, 10, 10], [2, 20, 20, 20, 20], [3, 10, 30, 20, 20], [4, 10, 40, 40, 40]]
insurance = pd.DataFrame(data, columns=['pid', 'tiv_2015', 'tiv_2016', 'lat', 'lon']).astype({'pid':'Int64', 'tiv_2015':'Float64', 'tiv_2016':'Float64', 'lat':'Float64', 'lon':'Float64'})
```

```python
df_insurance_1 = insurance[insurance['tiv_2015'].duplicated(keep=False)]
df_insurance_2 = insurance.drop_duplicates(subset=['lat','lon'],keep=False) 

df_insurance = df_insurance_1.merge(df_insurance_2,how='inner',left_on='pid',right_on='pid') 
df_insurance = df_insurance.filter(items={'tiv_2015_x','tiv_2016_x'})
df_insurance = df_insurance.rename(columns={'tiv_2015_x':'tiv_2015','tiv_2016_x':'tiv_2016'})


df_insurance = df_insurance.groupby('tiv_2015').sum().reset_index() # выдаст результат оп разным совпадающим продажам в 2015
df_insurance = df_insurance.filter(items={'tiv_2016'})
df_insurance['tiv_2016'] = df_insurance['tiv_2016'].sum() #просто суммирование и общая сумма
df_insurance = df_insurance.head(1)
#df_insurance['tiv_2016'] = df_insurance.groupby('tiv_2016').sum().reset_index()
#df_insurance['tiv_2016'] = df_insurance['tiv_2016'].sum() #просто суммирование и общая сумма
print(df_insurance)
```
