Parentheses (Klammern): Bearbeite zuerst alle Ausdrücke innerhalb von Klammern.
Exponents (Potenzen): Löse danach alle Potenzen und Wurzeln.
Multiplication und Division (Multiplikation und Division): Führe diese Operationen von links nach rechts aus, je nachdem, welche zuerst im Ausdruck vorkommt.
Addition und Subtraction (Addition und Subtraktion): Führe diese Operationen von links nach rechts aus, je nachdem, welche zuerst im Ausdruck vorkommt.
